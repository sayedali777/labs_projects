# Topic 1 : Predicting Customer Credit Score Category

## Problem Statement

Banks and lending institutions need a fast, consistent way to assess how creditworthy a customer is before approving loans, credit cards, or setting interest rates. Today this is often done through slow, manual review of a customer's financial history, which is inconsistent across reviewers and doesn't scale well as the number of applicants grows.

This project uses a customer's financial and credit history data (income, outstanding debt, payment history, number of loans/accounts, etc.) to predict which **Credit Score category** — *Poor*, *Standard*, or *Good* — a customer falls into.

## Goals

- Build a classification model that predicts a customer's Credit Score category (Poor / Standard / Good) from their financial profile.
- Identify which financial behaviors (e.g. late payments, outstanding debt, credit history length) most strongly drive a customer's credit standing.
- **Who benefits:** banks/lenders get a faster, more consistent screening tool, and the underlying insights can help customers understand what actions actually move their credit score.

## Approach

1. Clean the dataset — many numeric columns were exported as text with stray underscores and a few invalid values (e.g. negative ages), so these need to be converted to proper numeric types first.
2. Explore the data (this notebook): understand the target, feature types, missing values, correlations, and outliers.
3. Encode categorical features and handle missing values / outliers based on the EDA findings below.
4. Train and compare classification models (e.g. Logistic Regression, Random Forest, XGBoost), using metrics like accuracy and macro F1 (since the classes are imbalanced).

## Data Info

**Data Source:** [Credit Score Classification — Kaggle](https://www.kaggle.com/datasets/parisrohan/credit-score-classification?select=train.csv)

Rows: **100,000** | Columns: **28** (exceeds the 15k rows / 10 columns requirement)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv('Credit_score_classification.csv', low_memory=False)
print(df.shape)
df.head()
```

### Data Cleaning

Several numeric columns (`Age`, `Annual_Income`, `Num_of_Loan`, `Num_of_Delayed_Payment`, `Changed_Credit_Limit`,
`Outstanding_Debt`, `Amount_invested_monthly`, `Monthly_Balance`) were loaded as text because of stray trailing
underscores in the raw export (e.g. `"28_"`). We strip those and convert to numeric. We also convert
`Credit_History_Age` ("22 Years and 1 Months") into a single numeric `Credit_History_Age_Months` column, and treat
clearly invalid ages (negative or > 100) as missing.

```python
def clean_numeric(series):
    return pd.to_numeric(series.astype(str).str.replace('_', '', regex=False).str.strip(), errors='coerce')

messy_numeric_cols = ['Age', 'Annual_Income', 'Num_of_Loan', 'Num_of_Delayed_Payment',
                       'Changed_Credit_Limit', 'Outstanding_Debt', 'Amount_invested_monthly',
                       'Monthly_Balance']

for col in messy_numeric_cols:
    df[col] = clean_numeric(df[col])

# invalid ages
df.loc[(df['Age'] < 0) | (df['Age'] > 100), 'Age'] = np.nan

# Credit_History_Age -> total months
import re
def to_months(x):
    if pd.isna(x):
        return np.nan
    m = re.match(r'(\d+) Years and (\d+) Months', str(x))
    return int(m.group(1)) * 12 + int(m.group(2)) if m else np.nan

df['Credit_History_Age_Months'] = df['Credit_History_Age'].apply(to_months)

df[messy_numeric_cols + ['Credit_History_Age_Months']].describe()
```

### What is your target column and what does it represent?

The target is **`Credit_Score`** — a lender-assigned category (*Poor*, *Standard*, *Good*) summarizing how creditworthy
a customer currently is, based on their financial and repayment behavior.

### Is your target categorical or continuous? Is it balanced?

`Credit_Score` is **categorical** with 3 classes, so this is a **classification** problem. As the plot below shows,
the classes are **imbalanced**: *Standard* is the majority class (~53%), *Poor* is second (~29%), and *Good* is the
smallest (~18%). This imbalance means accuracy alone will be misleading later — macro F1 or a similar
balanced metric should be used to evaluate models.

```python
target_counts = df['Credit_Score'].value_counts()
print(target_counts)
print((target_counts / len(df) * 100).round(1))

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Credit_Score', order=target_counts.index, palette='viridis')
plt.title('Distribution of Credit_Score (target)')
plt.xlabel('Credit Score Category')
plt.ylabel('Count')
plt.show()
```

### What types of features do you have (categorical / numerical)? How many of each?

After cleaning, there are **17 numeric** columns and **12 categorical/text** columns (including identifier columns
like `ID`, `Customer_ID`, `Name`, `SSN` which are not useful predictive features and will be dropped before modeling).

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()

print(f"Numeric columns ({len(num_cols)}):", num_cols)
print()
print(f"Categorical/text columns ({len(cat_cols)}):", cat_cols)
```

### How many null values does each feature have? What's your plan for each?

Key columns with missing values after cleaning:

| Column | # Missing | Plan |
| --- | --- | --- |
| Monthly_Inhand_Salary | ~15,002 | Impute with median (per Occupation group) |
| Num_of_Delayed_Payment | ~7,002 | Impute with median |
| Credit_History_Age_Months | ~9,030 | Impute with median or per-customer forward/back fill (same customer appears across months) |
| Amount_invested_monthly | ~4,479 | Impute with median |
| Type_of_Loan | ~11,408 | Fill with "No Loan" category |
| Changed_Credit_Limit | ~2,091 | Impute with median |
| Num_Credit_Inquiries | ~1,965 | Impute with median |
| Age | ~2,776 (after removing invalid values) | Impute with median, or fill from other rows of the same Customer_ID |
| Monthly_Balance | ~1,200 | Impute with median |
| Name / SSN | ~9,985 / — | Drop — not predictive, identifier-like |

Since `Customer_ID` repeats across months for the same customer, a good imputation strategy is to first fill missing
values using that customer's other monthly records before falling back to a global median.

```python
null_counts = df.isnull().sum()
null_pct = (null_counts / len(df) * 100).round(2)
pd.DataFrame({'missing_count': null_counts, 'missing_pct': null_pct}).sort_values('missing_count', ascending=False).head(15)
```

### Which features are most correlated with your target, and with each other?

Encoding `Credit_Score` numerically (Poor=0, Standard=1, Good=2) to check linear relationships: the strongest
correlations with the target are `Delay_from_due_date` (**-0.43**), `Outstanding_Debt` (**-0.39**), and
`Credit_History_Age_Months` (**+0.39**) — i.e., customers who delay payments more and carry more debt tend to have
worse scores, while a longer credit history is associated with better scores.

```python
df_corr = df.copy()
df_corr['Credit_Score_num'] = df_corr['Credit_Score'].map({'Poor': 0, 'Standard': 1, 'Good': 2})

corr_with_target = df_corr[num_cols + ['Credit_Score_num']].corr()['Credit_Score_num'] \
    .drop('Credit_Score_num').sort_values(key=abs, ascending=False)
print(corr_with_target)

plt.figure(figsize=(10, 8))
sns.heatmap(df_corr[num_cols].corr(), cmap='coolwarm', center=0)
plt.title('Correlation matrix — numeric features')
plt.tight_layout()
plt.show()
```

### Are there outliers? How would you handle them?

Several financial columns (`Annual_Income`, `Num_Bank_Accounts`, `Num_Credit_Card`, `Interest_Rate`, `Num_of_Loan`,
`Num_of_Delayed_Payment`) have extreme outlier values that look like data-entry errors (e.g. thousands of bank
accounts or credit cards for one person). The boxplots below make these visible. Plan: cap values at a reasonable
domain limit (e.g. using the 1st/99th percentile or a hard business-rule cap) rather than dropping rows, since these
are likely corrupted values rather than genuinely extreme customers.

```python
outlier_check_cols = ['Annual_Income', 'Num_Bank_Accounts', 'Num_Credit_Card', 'Interest_Rate',
                      'Num_of_Loan', 'Num_of_Delayed_Payment', 'Outstanding_Debt']

fig, axes = plt.subplots(2, 4, figsize=(18, 8))
for ax, col in zip(axes.flatten(), outlier_check_cols):
    sns.boxplot(y=df[col], ax=ax, color='skyblue')
    ax.set_title(col)
axes[-1, -1].axis('off')
plt.tight_layout()
plt.show()

for col in outlier_check_cols:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {n_outliers} outlier rows (IQR bounds: {lower:.1f} to {upper:.1f})")
```
